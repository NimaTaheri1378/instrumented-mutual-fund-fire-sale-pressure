```python
from pathlib import Path
import pandas as pd

ROOT = Path.cwd().resolve()
while not (ROOT / 'configs' / 'project.yml').exists():
    ROOT = ROOT.parent
print(ROOT)
```

```python
import json
manifest = ROOT / 'manifests' / 'models_full.json'
if manifest.exists():
    models = json.loads(manifest.read_text())
    for key in ['elastic_net','lightgbm','xgboost','torch_mlp']:
        print(key, {k: models[key].get(k) for k in ['test_rank_ic','test_rank_icir','gpu_used'] if key in models})
```

```python
for name in ['model_horse_race_full.csv','lightgbm_feature_importance_full.csv','lightgbm_shap_summary_full.csv']:
    path = ROOT / 'reports' / 'tables' / name
    if path.exists():
        print('\n', name)
        display(pd.read_csv(path).head(20))
```
